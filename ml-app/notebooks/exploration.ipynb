# PredictIQ — Dataset Exploration

Run the cells in order. The notebook expects `input/car_data.csv` and `input/Titanic_train.csv` in the project root.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Find the project root whether the notebook is opened from the project or notebooks folder.
cwd = Path.cwd()
ROOT = cwd if (cwd / 'input').exists() else cwd.parent
INPUT_DIR = ROOT / 'input'
car_path = INPUT_DIR / 'car_data.csv'
titanic_path = INPUT_DIR / 'Titanic_train.csv'

print('Project root:', ROOT)
print('Car dataset exists:', car_path.exists())
print('Titanic dataset exists:', titanic_path.exists())
```

```python
car_df = pd.read_csv(car_path)
titanic_df = pd.read_csv(titanic_path)

print('Car dataset shape:', car_df.shape)
display(car_df.head())
print('Titanic dataset shape:', titanic_df.shape)
display(titanic_df.head())
```

```python
print('CAR DATA TYPES')
display(car_df.dtypes.to_frame('data type'))
print('CAR MISSING VALUES')
display(car_df.isna().sum().sort_values(ascending=False).to_frame('missing count'))
print('TITANIC DATA TYPES')
display(titanic_df.dtypes.to_frame('data type'))
print('TITANIC MISSING VALUES')
display(titanic_df.isna().sum().sort_values(ascending=False).to_frame('missing count'))
```

```python
display(car_df.describe(include='all').T)
display(titanic_df.describe(include='all').T)
```

```python
if 'Price' in car_df.columns:
    prices = pd.to_numeric(car_df['Price'], errors='coerce').dropna()
    plt.figure(figsize=(8, 4))
    plt.hist(prices, bins=30)
    plt.title('Car Price Distribution')
    plt.xlabel('Price')
    plt.ylabel('Number of cars')
    plt.tight_layout()
    plt.show()

if 'Survived' in titanic_df.columns:
    counts = titanic_df['Survived'].value_counts().sort_index()
    plt.figure(figsize=(6, 4))
    counts.plot(kind='bar')
    plt.title('Titanic Survival Counts')
    plt.xlabel('Survived (0 = no, 1 = yes)')
    plt.ylabel('Passengers')
    plt.tight_layout()
    plt.show()
```
